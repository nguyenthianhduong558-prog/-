```python
def square(n):
    return n ** 2
```

```python
def past(h, m, s):
    return (h * 3600 + m * 60 + s) * 1000
```

```python
def find_average(nums):
    if not nums:
        return 0
    return sum(nums) / len(nums)
```

```python
def is_square(n):
    if n < 0:
        return False
    root = int(n ** 0.5)
    return root * root == n
```

```python
def opposite(number):
    return -number
```
