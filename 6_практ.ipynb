```python
class Student:
    def __init__(self, name, group, tel_number):
        self.name = name
        self.group = group
        self.grades = []
        self.__tel_number = tel_number
    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        print(f"Имя: {self.name}")
        print(f"Группа: {self.group}")
        print(f"Оценки: {self.grades}")
        print(f"Средний балл: {self.get_average():.2f}")
        print(f"Телефон: {self.__tel_number}")


student1 = Student("Наталья Сидорова", "П-21", "+79000524267")
student2 = Student("Алексей Новиков", "П-22", "+78005553535")
student3 = Student("Ольга Драгунова", "П-21", "+79123456452")


student1.add_grade(5)
student1.add_grade(4)
student1.add_grade(4)

student2.add_grade(3)
student2.add_grade(4)
student2.add_grade(4)

student3.add_grade(5)
student3.add_grade(5)
student3.add_grade(4)


student1.info()
print(f"Средний балл Натальи: {student1.get_average():.2f}\n")

student2.info()
print(f"Средний балл Алексея: {student2.get_average():.2f}\n")

student3.info()
print(f"Средний балл Ольги: {student3.get_average():.2f}")
```
