```python
#1
stroka = input('Введите строку: ')
num = int(stroka)
obr = str(num)
print(f'Строка: {stroka}, тип: {type(stroka)}')
print(f'Число: {num}, тип: {type(num)}')
print(f'Обратно в строку: {obr}, тип: {type(obr)}')
```

```python
#2
n = int(input('Введите число: '))
otv = n if n < 0 else -n
print(otv)
```

```python
#3
cap = int(input('Вместимость автобуса (cap): '))
on = int(input('Пассажиров в автобусе (on): '))
wait = int(input('Ожидающих пассажиров (wait): '))
free_seats = cap - on
left = max(0, wait - free_seats)
print(left)
```
