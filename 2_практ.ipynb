```python
#1
def alarm(employed: bool, vacation: bool) -> bool:
  return employed and not vacation

print(alarm(True, True))
print(alarm(True, False))
print(alarm(False, True))
print(alarm(False, False))
```

```python
#2
def calculate(operation: str, value_1: int, value_2: int) -> int:
  if operation == '+':
    return value_1 + value_2
  elif operation == '-':
    return value_1 - value_2
  elif operation == '*':
    return value_1 * value_2
  elif operation == '/':
    return value_1 // value_2
  return 0

print(calculate('+', 4, 7))
print(calculate('-', 15, 18))
print(calculate('*', 5, 5))
print(calculate('/', 49, 7))
```

```python
#3
def summa(a: int, b: int) -> int:
  start, end = min(a, b), max(a, b)
  return sum(range(start, end + 1))

print(summa(1, 5))
print(summa(7, 3))
```

```python
#4
def kvart_season(month: int):
  if not(1 <= month <= 12):
    return 'Некорректный номер месяца'

  kvart = (month - 1) // 3 + 1
  if month in (12, 1, 2):
    season = 'зима'
  elif month in (3, 4, 5):
    season = 'весна'
  elif month in (6, 7, 8):
    season = 'лето'
  else:
    season = 'осень'

  return f'{kvart} квартал, {season}'

month = int(input('Введите номер месяца (1-12): '))
print(kvart_season(month))
```

```python
#5
def calculate_grade():
    scores = []

    while True:
        score = float(input("Введите оценку (для завершения введите -1): "))
        if score == -1:
            break
        scores.append(score)

    if not scores:
        return "Оценки не были введены"

    average = sum(scores) / len(scores)

    if average >= 90:
        return 'A'
    elif average >= 80:
        return 'B'
    elif average >= 70:
        return 'C'
    elif average >= 60:
        return 'D'
    else:
        return 'F'


result = calculate_grade()
print(f"Итоговая оценка: {result}")
```
