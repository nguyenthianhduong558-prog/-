```python
#1
n = int(input("Введите количество человек: "))

total = 0
for i in range(1, n + 1):
    age = int(input(f"Возраст человека {i}: "))
    total += age

average = total / n
print(f"Средний возраст: {average:.2f}")
```

```python
#2
code = input("Введите цветовой код (например a0351f): ").strip().lower()

if len(code) != 6 or not all(c in "0123456789abcdef" for c in code):
    print("Ошибка: код должен состоять из 6 шестнадцатеричных символов")
else:
    r = int(code[0:2], 16)
    g = int(code[2:4], 16)
    b = int(code[4:6], 16)

    if r == 0 and g == 0 and b == 0:
        print("черный")
    elif r == 255 and g == 255 and b == 255:
        print("белый")
    elif r == g == b:
        print("серый")
    else:
        if r >= g and r >= b:
            print("ближе к красному")
        elif g >= r and g >= b:
            print("ближе к зеленому")
        else:
            print("ближе к синему")
```
