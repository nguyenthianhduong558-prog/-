```python
#1
import json


def main():
    try:
        with open("students.json", "r", encoding="utf-8") as file:
            students = json.load(file)


        if not isinstance(students, list):
            raise TypeError("В students.json должен находиться список студентов")

        report = []
        all_grades = []

        for index, student in enumerate(students, start=1):
            if not isinstance(student, dict):
                raise TypeError(f"Запись №{index} должна быть словарём")

            if "name" not in student:
                raise KeyError(f"В записи №{index} отсутствует ключ 'name'")

            if "grades" not in student:
                raise KeyError(f"В записи №{index} отсутствует ключ 'grades'")

            name = student["name"]
            grades = student["grades"]


            if not isinstance(name, str):
                raise TypeError(f"Имя в записи №{index} должно быть строкой")

            if not isinstance(grades, list):
                raise TypeError(f"Оценки студента {name} должны быть списком")

            if not grades:
                raise ValueError(f"У студента {name} нет оценок")

            if not all(isinstance(grade, (int, float)) and not isinstance(grade, bool)
                       for grade in grades):
                raise TypeError(f"Оценки студента {name} должны быть числами")

            average = sum(grades) / len(grades)
            all_grades.extend(grades)

            report.append({
                "name": name,
                "average_grade": round(average, 2)
            })

        overall_average = sum(all_grades) / len(all_grades) if all_grades else 0

        result = {
            "students": report,
            "overall_average": round(overall_average, 2)
        }


        with open("report.json", "w", encoding="utf-8") as file:
            json.dump(result, file, ensure_ascii=False, indent=4)

        print("Отчет успешно сохранён в report.json")

    except FileNotFoundError:
        print("Ошибка: файл students.json не найден.")
    except json.JSONDecodeError:
        print("Ошибка: students.json содержит некорректный JSON.")
    except (KeyError, TypeError, ValueError) as error:
        print(f"Ошибка в данных: {error}")
    except OSError as error:
        print(f"Ошибка при работе с файлом: {error}")


if __name__ == "__main__":
    main()
```
