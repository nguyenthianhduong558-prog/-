```python
#1
#2 задание
def calculate(operation: str, value_1: int, value_2: int) -> int:
  if operation == '+':
    return value_1 + value_2
  elif operation == '-':
    return value_1 - value_2
  elif operation == '*':
    return value_1 * value_2
  elif operation == '/':
    return value_1 // value_2
  return 0

print(calculate('+', 4, 7))
print(calculate('-', 15, 18))
print(calculate('*', 5, 5))
print(calculate('/', 49, 7))

#4 задание
def kvart_season(month: int):
  if not(1 <= month <= 12):
    return 'Некорректный номер месяца'

  kvart = (month - 1) // 3 + 1
  if month in (12, 1, 2):
    season = 'зима'
  elif month in (3, 4, 5):
    season = 'весна'
  elif month in (6, 7, 8):
    season = 'лето'
  else:
    season = 'осень'

  return f'{kvart} квартал, {season}'

month = int(input('Введите номер месяца (1-12): '))
print(kvart_season(month))
```

```python
#2
def count_target_chars(text: str, target_chars: list) -> dict:
    result = {char: 0 for char in target_chars}

    for char in text:
        if char in result:
            result[char] += 1
    return result


s = "I like ice cream and cheesecake"
target_list = ["i", "l", "c", "e"]

char_counts = count_target_chars(s, target_list)


for char, count in char_counts.items():
    print(f"Символ '{char}': {count}")
```
