```python
#1
strings_list = ["круг", "квадрат", "треугольник", "ромб"]
target_string = input("Введите строку для поиска: ")

if target_string in strings_list:
    index = strings_list.index(target_string)
    print(f"Строка найдена на индексе: {index}")
else:
    print("Строка не найдена в списке.")
```

```python
#2
original_list = [1, 3, 2, 3, 4, 1, 5]
unique_list = list(set(original_list))

print("Результат:", unique_list)
```

```python
#3
items = ["вишня", "черешня", "вишня", "морошка", "черешня", "вишня"]
counts = {}

for item in items:
    counts[item] = counts.get(item, 0) + 1

print("Количество повторений:", counts)
```

```python
#4
num = [1488, 3, 67, 2, 8, 2, 52]

if num:
    min_val = min(num)
    num.remove(min_val)

print("Список после удаления минимального числа:", num)
```
