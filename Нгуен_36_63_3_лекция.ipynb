```python
def maximum(numbers):
    return max(numbers)

def minimum(numbers):
    return min(numbers)
```

```python
def name_value(my_list):
    result = []
    for position, word in enumerate(my_list, start=1):
        value = sum(ord(ch) - ord('a') + 1 for ch in word if ch != ' ')
        result.append(value * position)
    return result
```

```python
class List:
    @staticmethod
    def remove_(integer_list, values_list):
        values_set = set(values_list)
        return [x for x in integer_list if x not in values_set]
```

```python
def remove_duplicate_words(s):
    seen = set()
    result = []
    for word in s.split():
        if word not in seen:
            seen.add(word)
            result.append(word)
    return ' '.join(result)


text = 'alpha beta beta gamma gamma gamma delta alpha beta beta gamma gamma gamma delta'
print(remove_duplicate_words(text))
```
